This notebook implements a multi-level compression system that:
- Preserves decision-critical content (numbers, dates, exceptions, risks)

## Architecture:
```
Paragraph Level → Section Level → Chapter Level → Document Level
     ↓                ↓                ↓                ↓
  GETS +          BART +           BART +           BART +
  NER +           Detect           Summarize        Summarize
  Regex           Contradictions
```

## Step 4: Run the Compression Pipeline

This will:
1. Extract text from PDF (paragraph-level)
2. Process each paragraph (GETS + NER + Regex)
3. Build hierarchical structure (Paragraph → Section → Chapter → Document)
4. Generate summaries at each level using BART
5. Detect contradictions
6. Save results to JSON

````python
from contextual_compression_engine import HierarchicalCompressor
import json

# Initialize the compressor
print("Initializing Contextual Compression Engine...")
compressor = HierarchicalCompressor()

# Extract text from PDF
print(f"\nProcessing: {pdf_filename}")
paragraphs = compressor.extract_text_from_pdf(pdf_filename)

# Perform hierarchical compression
# Adjust section_size and chapter_size based on document length
section_size = 5  # Paragraphs per section
chapter_size = 3  # Sections per chapter

print(f"\nConfiguration:")
print(f"  - Section size: {section_size} paragraphs")
print(f"  - Chapter size: {chapter_size} sections")
print(f"\nStarting compression...\n")

result = compressor.compress_hierarchically(
    paragraphs, 
    section_size=section_size, 
    chapter_size=chapter_size
)

# Save to JSON
output_json = 'compression_result.json'
with open(output_json, 'w') as f:
    json.dump(result, f, indent=2, default=str)

print(f"\n✓ Compression complete! Results saved to {output_json}")
````

## Step 5: View Compression Statistics

````python
import json

with open('compression_result.json', 'r') as f:
    result = json.load(f)

metadata = result['metadata']

print("="*60)
print("COMPRESSION STATISTICS")
print("="*60)
print(f"\nHierarchical Structure:")
print(f"  📄 Total Paragraphs:    {metadata['total_paragraphs']}")
print(f"  📑 Total Sections:      {metadata['total_sections']}")
print(f"  📚 Total Chapters:      {metadata['total_chapters']}")
print(f"\nDecision-Critical Content:")
print(f"  ⚠️  Critical Facts:      {metadata['total_facts']}")
print(f"  ⚡ Contradictions:      {metadata['total_contradictions']}")
print(f"\nCompression Ratio:")

# Calculate compression
original_length = sum(len(p['original_text']) for p in result['paragraphs'])
compressed_length = len(result['document_summary']['abstractive_summary'])
ratio = (1 - compressed_length / original_length) * 100 if original_length > 0 else 0

print(f"  📊 Original length:     {original_length:,} characters")
print(f"  📊 Compressed length:   {compressed_length:,} characters")
print(f"  📊 Compression ratio:   {ratio:.1f}%")
print("="*60)
````

## Step 11: Query Specific Content (Optional)

Search for specific facts or content in the compressed document.

````python
def search_facts(query, facts, top_k=5):
    """Simple keyword-based search for critical facts"""
    query_lower = query.lower()
    matches = []
    
    for fact in facts:
        content_lower = fact['content'].lower()
        context_lower = fact['context'].lower()
        
        if query_lower in content_lower or query_lower in context_lower:
            matches.append(fact)
    
    return matches[:top_k]

# Example query
query = input("Enter search term (e.g., 'limit', 'date', 'risk'): ")

matches = search_facts(query, result['critical_facts'])

print(f"\nFound {len(matches)} matches for '{query}':\n")
print("="*60)

for i, fact in enumerate(matches, 1):
    print(f"\n{i}. Type: {fact['fact_type']}")
    print(f"   Content: {fact['content']}")
    print(f"   Context: {fact['context'][:100]}...")
    print(f"   Source: Page {fact['source']['page']}, Para {fact['source']['paragraph_idx']}")

print("="*60)
````

## Step 12: Explainability - What Was Preserved vs. Removed?

Compare original paragraphs with their compressed versions.

````python
import random

# Pick a random paragraph to analyze
sample_para = random.choice(result['paragraphs'])

print("EXPLAINABILITY EXAMPLE")
print("="*60)
print(f"\nParagraph ID: {sample_para['id']}")
print(f"Source: Page {sample_para['source']['page']}, Para {sample_para['source']['paragraph_idx']}")
print("\n" + "-"*60)
print("ORIGINAL TEXT:")
print("-"*60)
print(sample_para['original_text'])
print("\n" + "-"*60)
print("EXTRACTIVE SUMMARY (GETS):")
print("-"*60)
print(sample_para['extractive_summary'])
print("\n" + "-"*60)
print("CRITICAL FACTS EXTRACTED:")
print("-"*60)

para_facts = [f for f in result['critical_facts'] 
              if f['source']['page'] == sample_para['source']['page'] 
              and f['source']['paragraph_idx'] == sample_para['source']['paragraph_idx']]

if para_facts:
    for fact in para_facts:
        print(f"  • {fact['fact_type']}: {fact['content']}")
else:
    print("  (No critical facts in this paragraph)")

print("\n" + "="*60)
print("\nWhat was preserved:")
print("  ✓ Key sentences (via GETS graph-based selection)")
print(f"  ✓ {len(para_facts)} critical facts (numbers, dates, exceptions, etc.)")
print("  ✓ Full source reference for traceability")
print("\nWhat was removed:")
print("  ✗ Non-critical descriptive text")
print("  ✗ Redundant phrasing")
print("  ✗ Filler content")
print("="*60)
````
